## Del 2: Lister og arrays

En **liste** kan holde flere værdier. Men se, hvad der sker, når man ganger en liste med 2: den bliver **gentaget**, ikke regnet på.

```python
# En liste med tre af Ishusets priser
priser_liste = [35, 45, 50]
print(priser_liste * 2)
```

Til beregninger bruger vi i stedet et **numpy-array**. Ganger man et array med 2, ganges **hvert tal**. Det er hele pointen med numpy: regning på mange tal på én gang.

```python
# Samme tal som array: nu regnes der
priser = np.array([35, 45, 50])
print(priser * 2)
```

**Indeksering og udsnit.** Python tæller fra 0: `priser[0]` er det første tal. Et udsnit `priser[0:2]` giver elementerne fra plads 0 **til og med** plads 1 (slutgrænsen er ikke med). Negative indeks tæller bagfra: `priser[-1]` er det sidste tal, `priser[-2]` det næstsidste.

```python
# Enkeltelementer og udsnit
print(priser[0])
print(priser[2])
print(priser[0:2])
print(priser[-1])    # det sidste tal
print(priser[-2:])   # de to sidste tal
```

**Elementvis regning.** Alt, man kan med ét tal, kan man med et helt array: gange, lægge til, trække fra. To lige lange arrays kan også regnes sammen, element for element.

```python
# Alle priser med 20 procent rabat
rabatpriser = priser * 0.8
print(rabatpriser)

# Prisstigning: 3 kr. oven i alle priser
print(priser + 3)

# To arrays lagt sammen, element for element
print(priser + rabatpriser)
```

**Tabeller: 2D-arrays.** Et array kan have rækker og kolonner, som et lille regneark. Her: solgte is pr. uge i de tre sommerkiosker. `salg[0]` er første række, `salg[:, 0]` er første kolonne, `salg[1, 2]` er ét bestemt tal (række 1, kolonne 2). Regning virker også på en enkelt række.

```python
# Rækker: Blokhus, Løkken, Skagen. Kolonner: uge 1 til 4.
salg = np.array([
    [120, 145, 160, 150],   # Blokhus
    [ 95, 110, 125, 115],   # Løkken
    [ 80, 100, 105,  95],   # Skagen
])

print(salg[0])       # hele Blokhus-rækken
print(salg[:, 0])    # uge 1 for alle tre kiosker
print(salg[1, 2])    # Løkken, uge 3
print(salg[0] * 2)   # hvis Blokhus solgte dobbelt
```

## Del 3: Funktioner og plots

**Egne funktioner med `def`.** En funktion tager et input og giver et output, som i matematik. Her: de daglige omkostninger for en kiosk, 400 kr. i faste udgifter plus 12 kr. pr. solgt is.

```python
# Daglige omkostninger som funktion af antal solgte is
def omkostning(x):
    return 400 + 12 * x

print(omkostning(0))
print(omkostning(10))
print(omkostning(100))
```

**Mange x-værdier på én gang: `np.linspace`.** `np.linspace(0, 40, 100)` giver 100 jævnt fordelte tal fra 0 til 40. Sender man dem gennem funktionen, får man 100 y-værdier: alt hvad et plot behøver.

```python
# 100 x-værdier fra 0 til 40 solgte is
x = np.linspace(0, 40, 100)
y = omkostning(x)

print(x[:5])   # de første fem x-værdier
print(y[:5])   # de tilhørende y-værdier
```

**Mønstret bag ethvert funktionsplot:** linspace, beregn, plot, tekst på akserne, overskrift. Det er de samme fem linjer hver gang, kun indholdet skifter.

```python
# Det fulde plotmønster
plt.plot(x, y)
plt.xlabel("Antal solgte is")
plt.ylabel("Kroner")
plt.title("Ishuset: daglige omkostninger")
plt.show()
```

## Del 4: Prikprodukt og udblik

**Prikprodukt: Excels SUMPRODUKT i Python.** En familie bestiller 2 stk. Vaffel 2 kugler, 1 softice og 1 slushice. Regningen er antal gange pris, lagt sammen. Først i to trin, så med genvejen `@`, der gør begge dele på én gang.

```python
# Familiens bestilling
antal  = np.array([2, 1, 1])
priser = np.array([50, 45, 35])

# Trin for trin: elementvis produkt, derefter sum
print(antal * priser)
print((antal * priser).sum())

# Genvejen: prikprodukt med @
print(antal @ priser)
```
